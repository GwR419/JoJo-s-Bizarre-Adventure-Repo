## Word Count for XML Files

This Python code is designed to output word frequency for our XML files. This will help us compare the anime to the mangna.



```python
import xml.etree.ElementTree as ET
from collections import Counter
import re
```

```python
tree = ET.parse("../xmlCorpus/04/04x37_-_King_of_Kings.xml")  # replace with the actual filename
root = tree.getroot()
```

```python
all_text = ""

for dialogue in root.findall("dialogue"):
    speech = dialogue.find("speech")
    if speech is not None and speech.text:
        all_text += " " + speech.text

print(all_text)  # make sure your text looks right
```

```python
# Get rid of unwanted words, not useful for our analysis
stop_words = {"the", "to", "you", "it", "is", "this", "of", "that", "a", "be", "i", "s", "t", "my", "what", "but", "me", "can", "no", "he", "was", "will", "and", "in", "for", "we", "are", "not", "so", "do", "have", "they", "at", "his", "her", "on", "or", "with", "as", "an", "if", "all", "by"}

# Lowercase everything, remove punctuation
words = re.findall(r"\b[a-z][a-z']*\b", all_text.lower())

# Count each word
word_counts = Counter(words)

# Looks for junk words classified in stop_words
filtered_words = [w for w in words if w not in stop_words and len(w) >= 3]
word_counts = Counter(filtered_words)

# Print every word and how many times it appears, sorted most -> least
for word, count in word_counts.most_common(30):
    print(f"{word}: {count}")
```
